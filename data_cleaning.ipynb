```python
import pandas as pd

# STEP 1 — LOAD

df = pd.read_csv("C:\\Users\\Greesha Vaishnavi\\Downloads\\customer_analytics - customer_analytics.csv")


# STEP 2 — INSPECT

df.head()
df.tail()
df.sample(10)
df.info()


# STEP 3 — MISSING VALUE ANALYSIS

df.isnull().sum()
df.isna().sum()

# STEP 4 — DUPLICATE HANDLING

df.duplicated().sum()

df = df.drop_duplicates()

df.duplicated().sum()


# STEP 5 — NUMERIC CLEANING

# Age
df['Age'] = pd.to_numeric(
    df['Age'],
    errors='coerce'
)

df['Age'] = df['Age'].fillna(
    df['Age'].median()
)


# Annual Income
df['Annual_Income'] = pd.to_numeric(
    df['Annual_Income'],
    errors='coerce'
)

df['Annual_Income'] = df['Annual_Income'].fillna(
    df['Annual_Income'].median()
)

df['Annual_Income'] = df['Annual_Income'].astype(float)


# Account Balance
df['Account_Balance'] = pd.to_numeric(
    df['Account_Balance'],
    errors='coerce'
)

df['Account_Balance'] = df['Account_Balance'].fillna(
    df['Account_Balance'].mean()
)


# Credit Score
df['Credit_Score'] = pd.to_numeric(
    df['Credit_Score'],
    errors='coerce'
)

df['Credit_Score'] = df['Credit_Score'].fillna(
    df['Credit_Score'].median()
)

df['Credit_Score'] = df['Credit_Score'].astype(int)


# Monthly Spend
df['Monthly_Spend'] = pd.to_numeric(
    df['Monthly_Spend'],
    errors='coerce'
)

df['Monthly_Spend'] = df['Monthly_Spend'].fillna(
    df['Monthly_Spend'].median()
)


# Transactions
df['Transactions'] = pd.to_numeric(
    df['Transactions'],
    errors='coerce'
)

df['Transactions'] = df['Transactions'].fillna(
    df['Transactions'].median()
)

df['Transactions'] = df['Transactions'].astype(int)


# Satisfaction Score
df['Satisfaction_Score'] = pd.to_numeric(
    df['Satisfaction_Score'],
    errors='coerce'
)

df['Satisfaction_Score'] = df['Satisfaction_Score'].fillna(
    df['Satisfaction_Score'].mean()
)

df['Satisfaction_Score'] = df['Satisfaction_Score'].astype(float)


# STEP 6 — DATE CLEANING

df['Join_Date'] = pd.to_datetime(
    df['Join_Date'],
    errors='coerce'
)

df['Join_Date'] = df['Join_Date'].fillna(
    df['Join_Date'].median()
)


# STEP 7 — CATEGORICAL CLEANING

# City
df['City'] = df['City'].replace({
    'mumbai': 'Mumbai',
    'hyderabad': 'Hyderabad',
    'delhi': 'Delhi',
    'pune': 'Pune',
    'chennai': 'Chennai',
    'bangalore': 'Bangalore',
    'kolkata': 'Kolkata'
})


# Segment
df['Segment'] = df['Segment'].fillna('Premium')


# Channel
df['Channel'].value_counts()


# Active
df['Active'] = df['Active'].replace({
    'Y': 'Yes',
    'YES': 'Yes',
    'yes': 'Yes',
    1: 'Yes',
    'N': 'No',
    'NO': 'No',
    'no': 'No',
    0: 'No'
})

df['Active'] = df['Active'].fillna('No')


# Risk Level
df['Risk_Level'].value_counts()


# STEP 8 — COLUMN TRANSFORMATION

df = df.rename(columns={
    'Customer_ID': 'Customer_ID',
    'Annual_Income': 'AnnualIncome',
    'Account_Balance': 'AccountBalance',
    'Credit_Score': 'CreditScore',
    'Tenure_Years': 'Tenure',
    'Join_Date': 'CustomerJoinDate',
    'Risk_Level': 'RiskLevel',
    'Monthly_Spend': 'MonthlySpend',
    'Satisfaction_Score': 'SatisfactionScore'
})


df = df.add_prefix('Customer_')


df = df.drop(
    'Customer_Customer_ID',
    axis=1
)

# STEP 9 — ANALYTICAL FEATURES

df['Income_Lakhs'] = (
    df['Customer_AnnualIncome'] / 100000
)

df['MonthlyIncome'] = (
    df['Customer_AnnualIncome'] / 12
)

df['Income_After_Spending'] = (
    df['MonthlyIncome'] -
    df['Customer_MonthlySpend']
)

df['Balance_Income_Ratio'] = (
    df['Customer_AccountBalance'] /
    df['Customer_AnnualIncome'].replace(0, pd.NA)
)

df['Customer_City'] = df['Customer_City'].fillna('Unknown')

df['Customer_Active'] = df['Customer_Active'].replace({
    1: 'Yes',
    0: 'No',
    '1': 'Yes',
    '0': 'No'
})


df['Customer_City'].isna().sum()

df['Customer_Active'].value_counts()

# STEP 10 — FINAL VALIDATION

# Duplicate check
print("Duplicates:")
print(df.duplicated().sum())


# Datatypes
print("\nDatatypes:")
print(df.dtypes)


# Missing values
print("\nMissing Values:")
print(df.isnull().sum())


# Numeric statistics
print("\nDescriptive Statistics:")
print(df.describe())


# Categorical distributions
print("\nCity:")
print(df['Customer_City'].value_counts())


print("\nSegment:")
print(df['Customer_Segment'].value_counts())


print("\nChannel:")
print(df['Customer_Channel'].value_counts())


print("\nActive:")
print(df['Customer_Active'].value_counts())


print("\nRisk Level:")
print(df['Customer_RiskLevel'].value_counts())


# Final inspection
print("\nFirst 10 rows:")
print(df.head(10))

print("\nLast 10 rows:")
print(df.tail(10))

print("\nRandom 10 rows:")
print(df.sample(10))

print("\nFinal Information:")
df.info()

df['Customer_City'] = df['Customer_City'].fillna('Unknown')
```
